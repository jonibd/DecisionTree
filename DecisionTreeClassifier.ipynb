```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier,plot_tree
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
from sklearn.datasets import load_iris
from sklearn.ensemble import RandomForestClassifier
```

```python
iris=load_iris()
```

```python
df=pd.DataFrame(data=iris.data,columns=iris.feature_names)
df['target']=iris.target
```

```python
df.head()
```

```python
df.shape
```

```python
df['target']
```

```python
x=df.drop('target',axis=1)
y=df['target']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=.20,random_state=42)
```

```python
model=DecisionTreeClassifier(max_depth=3, min_samples_split=20,min_samples_leaf=10,max_features=None)
```

```python
model.fit(x_train, y_train)
```

```python
plt.figure(figsize=(10,8))
plot_tree(model,feature_names=iris.feature_names,class_names=iris.target_names,filled=True, rounded=True)
plt.show()
```

```python
y_prad=model.predict(x_test)
```

```python
y_prad
```

```python
accuracy=accuracy_score(y_test,y_prad)
```

```python
accuracy
```

```python
confutions_m=confusion_matrix(y_test,y_prad)
```

```python
confutions_m
```

```python
classification_r=classification_report(y_test,y_prad)
```

```python
print(classification_r)
```

```python
plot_tree(model1.estimators_[0],
          feature_names=iris.feature_names,
          class_names=iris.target_names,
          filled=True,
          rounded=True)

plt.show()
```
