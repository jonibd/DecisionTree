```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
```

```python

data = {
    'Age': [25, 30, 45, 35, 22, 40],
    'Income': [30, 40, 60, 50, 25, 70],
    'Loan': ['Yes', 'Yes', 'Yes', 'No', 'No', 'Yes']
}

df = pd.DataFrame(data)
df
```

```python
x=df[['Age','Income']]
y=df['Loan']
```

```python
model=DecisionTreeClassifier(criterion='gini')
```

```python
model.fit(x,y)
```

```python
model.predict([[28, 35]])
```

```python
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))
plot_tree(model, feature_names=['Age', 'Income'], class_names=['No', 'Yes'], filled=True)
plt.show()
```

```python
model = DecisionTreeClassifier(criterion='gini', max_depth=1, min_samples_leaf=3)
```

```python
model.fit(x, y)
```

```python
plt.figure(figsize=(10,6))
plot_tree(model, feature_names=['Age','Income'], class_names=['No','Yes'], filled=True)
plt.show()
```
